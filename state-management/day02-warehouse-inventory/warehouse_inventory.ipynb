```python
inventory = {
    "keyboard": 5,
    "mouse": 3,
    "monitor": 2,
}
```

```python
events = [
    ("reserve", "customer_a", "keyboard", 2),
    ("reserve", "customer_b", "keyboard", 4),
    ("release", "customer_a", "keyboard", 1),
    ("reserve", "customer_b", "keyboard", 3),
    ("purchase", "customer_a", "keyboard", 1),
]
```

```python
def warehouse_inventory(inventory, events):

    reservations = {}
    results = []
    for action, customer_id, item, quantity in events:
        # status of rejection at base case for conditionals
        status = "rejected"

        if customer_id not in reservations:
            # create the subdictionary by customer_id
            reservations[customer_id] = {}
            if item not in reservations[customer_id]:
                # initialize the item counter
                reservations[customer_id][item] = 0
        reserved_quantity = reservations[customer_id][item]
        if action == "reserve":
            if item in inventory and inventory[item] >=quantity:
                inventory[item] -= quantity
                reservations[customer_id][item] += quantity
                status = "accepted"
        elif action == "purchase":
            if reserved_quantity >= quantity:
                reservations[customer_id][item] -= quantity
                status = "accepted"

        elif action == "release":
            inventory[item] += quantity
            reservations[customer_id][item] -= quantity
            status = "accepted"

        results.append((
            action,
            customer_id,
            item,
            quantity,
            status,
        ))
    return results


```

```python
warehouse_inventory(inventory, events)
```
