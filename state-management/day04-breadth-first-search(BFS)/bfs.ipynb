1. Algorithm needs to maintain a queue of nodes waiting to be searched, a set of visited nodes, and the path used to reach each queued node
2. Use a FIFO queue, implemented in Ptyhon with `collections.deque`
3. Invariants - each node must be queued once, and nodes are processed in order of increasing distance from the start
4. USes explicit FIFO queue (`popleft`)

```python
# given this graph

graph = {
    "A": ["B", "C"],
    "B": ["D", "E"],
    "C": ["F"],
    "D": [],
    "E": ["F"],
    "F": []
}
```

```python
from collections import deque


def shortest_path(graph, start, target):
    next_to_go = deque()
    visited = set()

    next_to_go.append((start, [start]))
    visited.add(start)

    while next_to_go:
        current_node, path = next_to_go.popleft()


        print("Current:", current_node)
        print("Path:", path)
        print("Queue:", next_to_go)


        if current_node == target:
            return path



        for neighbor in graph[current_node]:
            if neighbor not in visited:
                visited.add(neighbor)
                new_path = path + [neighbor]
                print (f'new path: {new_path}, neighbor: {neighbor}')
                next_to_go.append((neighbor, new_path))
    return None









```

```python
shortest_path(graph, "A", "F")
```
