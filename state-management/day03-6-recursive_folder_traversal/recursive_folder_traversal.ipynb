```python
folders = {
    "projects": { # dictionary represents a folder
        "ai": {
            "agent.py": None, # None represents a file
            "evals.py": None,
        },
        "web": {
            "app.py": None,
        },
    },
    "notes.txt": None,
}
```

```python
def print_files(folder, depth=0):
    for name, contents in folder.items():
        print("     " * depth + name)

        if contents is not None:
            print_files(contents, depth + 1)


print_files(folders)
```

```python
def count_files(folder):
    total_count = 0
    for name, contents in folder.items():
        if contents is None:
            total_count +=1
        else:
            total_count +=count_files(contents)
    return total_count
    
```

```python
count_files(folders)
```

```python
folders = {
    "projects": {
        "ai": {
            "agent.py": 1200,
            "evals.py": 800,
        },
        "web": {
            "app.py": 2500,
        },
    },
    "notes.txt": 300,
}
```

```python
def file_size(folder):
    byte_size = 0
    for name, contents in folder.items():
        if isinstance(contents, int):
            byte_size += contents

        else:
            byte_size += file_size(contents)


    return byte_size
```

```python
file_size(folders)
```
